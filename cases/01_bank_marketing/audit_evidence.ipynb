# Audit evidence: Bank Marketing

The audit reproduces the saved random 60/20/20 split, then compares it with a 60/20/20 split by source row order. The source has month and weekday fields but no year or complete timestamp, so row order is only a temporal stress test. The code displays aggregate metrics only and persists the computed results and report.

```python
import json
import os
import sys
from pathlib import Path

project_root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'src' / 'bank_audit.py').is_file())
os.chdir(project_root)
sys.path.insert(0, str(project_root / 'src'))
from bank_audit import audit_bank_marketing, render_bank_audit_markdown, write_bank_audit_results

audit = audit_bank_marketing()
results_path = write_bank_audit_results(audit)
report_path = render_bank_audit_markdown(results_path)
evidence = {
    'random_split': audit['random_split'],
    'source_order_split': audit['source_order_split'],
    'source_order_split_definition': audit['source_order_split_definition'],
    'test_split_gaps_source_order_minus_random': audit['test_split_gaps_source_order_minus_random'],
    'source_order_feature_ablations': audit['source_order_feature_ablations'],
    'data_and_feature_checks': audit['data_and_feature_checks'],
}
print('Aggregate evidence (source rows and individual predictions are not displayed)')
print(json.dumps(evidence, indent=2, allow_nan=False))
print(f'Updated {results_path.relative_to(project_root)} and {report_path.relative_to(project_root)}')
```
