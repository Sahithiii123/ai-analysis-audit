# Audit evidence: Telco Churn

The saved AI run fails while parsing the ARFF string attribute and emits no results. This notebook preserves that run and independently evaluates the descriptive claims through the repository loader. It also measures the singularity in the submitted logistic design and fits a corrected model after resolving exact duplicate service-category indicators. Outputs contain aggregates only.

```python
import json
import os
import sys
from pathlib import Path

project_root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'src' / 'telco_audit.py').is_file())
os.chdir(project_root)
sys.path.insert(0, str(project_root / 'src'))
from telco_audit import audit_telco_churn, render_telco_audit_markdown, write_telco_audit_results

audit = audit_telco_churn()
results_path = write_telco_audit_results(audit)
report_path = render_telco_audit_markdown(results_path)
evidence = {
    'saved_ai_run': audit['saved_ai_run'],
    'descriptive': audit['descriptive'],
    'mutual_information': audit['mutual_information'],
    'submitted_adjusted_model': audit['submitted_adjusted_model'],
    'corrected_adjusted_model': audit['corrected_adjusted_model'],
}
print('Aggregate audit evidence (no customer-level rows are displayed)')
print(json.dumps(evidence, indent=2, allow_nan=False))
print(f'Updated {results_path.relative_to(project_root)} and {report_path.relative_to(project_root)}')
```
